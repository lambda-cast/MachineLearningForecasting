# PV Forecasting — 1. Data Loading & Overview

**Purpose:** load per-month CSV files, concatenate them into a single time-ordered DataFrame, and get a first look at shape and summary statistics.

**Input:** `../Input/merged_2026-*.csv`

**Output:** `../Output/checkpoint_01_loaded.pkl` (used by the next notebook) and `Output/output.xlsx` (for manual inspection)

## Import libraries

```python
import pandas as pd
import glob
import os
```

## Configuration

```python
INPUT_DIR = "../Input/merged_2026-*.csv"
OUTPUT_DIR = "../Output"

os.makedirs(OUTPUT_DIR, exist_ok=True)  # Avoid FileNotFoundError on first run
```

## 1. Load & concatenate

```python
files = sorted(glob.glob(INPUT_DIR))
print(f"Found {len(files)} file(s):")

dfs = []
for f in files:
    d = pd.read_csv(f, parse_dates=["Timestamp"])
    dfs.append(d)
df = pd.concat(dfs, ignore_index=True).sort_values("Timestamp").reset_index(drop=True)
print(f"\nCombined shape: {df.shape}")
```
